```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

# Process FC_05_formation_0.csv

```python
df = pd.read_csv("FC_05_formation_0.csv")
```

```python
plt.plot(df["time/s"]/3600, df["Ecell/V"])
plt.plot(df["time/s"]/3600, df["mode"])
```

```python
mode_diff = df["mode"].diff()
index_switch = mode_diff[mode_diff != 0].index.to_list()
index_switch = [0] + [i -1 for i in index_switch[1:]] + [len(df["mode"])-1]

plt.plot(df["time/s"], df["Ecell/V"])

for i in index_switch:
    plt.axvline(x=df["time/s"].iloc[i], color="r")
```

```python
A_cell = np.pi * 1.48 ** 2 / 4
# capacity first charge
capacity_1ch = (df["(Q-Qo)/mA.h"].iloc[index_switch[2]] - df["(Q-Qo)/mA.h"].iloc[index_switch[0]]) / A_cell

# capacity first discharge
capacity_1dch = (df["(Q-Qo)/mA.h"].iloc[index_switch[3]] - df["(Q-Qo)/mA.h"].iloc[index_switch[4]]) / A_cell

# capacity second charge
capacity_2ch = (df["(Q-Qo)/mA.h"].iloc[index_switch[7]] - df["(Q-Qo)/mA.h"].iloc[index_switch[5]]) / A_cell

# capacity second discharge
capacity_2dch = (df["(Q-Qo)/mA.h"].iloc[index_switch[8]] - df["(Q-Qo)/mA.h"].iloc[index_switch[9]]) / A_cell
```

```python
print("Charge capacity cycle 1:", capacity_1ch)
print("Discharge capacity cycle 1:", capacity_1dch)
print("Charge capacity cycle 2:", capacity_2ch)
print("Discharge capacity cycle 2:", capacity_2dch)

# Coulombic efficiency: discharge capacity / charge capacity
print("Coulombic efficiency cycle 1:", capacity_1dch / capacity_1ch)
print("Coulombic efficiency cycle 2:", capacity_2dch / capacity_2ch)
```

# Process FC_05_cycling_1.csv

```python
df = pd.read_csv("FC_05_cycling_1.csv")
```

```python
mode_diff = df["mode"].diff()
index_switch = mode_diff[mode_diff != 0].index.to_list()
index_switch = [0] + [i -1 for i in index_switch[1:]] + [len(df["mode"])-1]

print(np.shape(index_switch))
plt.plot(df["time/s"], df["Ecell/V"])

for i in index_switch:
    plt.axvline(x=df["time/s"].iloc[i], color="r")


# plt.xlim([340000, 400000])
```

```python
A_cell = np.pi * 1.48 ** 2 / 4
# capacity third charge
capacity_3ch = (df["(Q-Qo)/mA.h"].iloc[index_switch[2]] - df["(Q-Qo)/mA.h"].iloc[index_switch[0]]) / A_cell

# capacity third discharge
capacity_3dch = (df["(Q-Qo)/mA.h"].iloc[index_switch[3]] - df["(Q-Qo)/mA.h"].iloc[index_switch[4]]) / A_cell

# capacity fourth charge
capacity_4ch = (df["(Q-Qo)/mA.h"].iloc[index_switch[7]] - df["(Q-Qo)/mA.h"].iloc[index_switch[5]]) / A_cell

# capacity fourth discharge
capacity_4dch = (df["(Q-Qo)/mA.h"].iloc[index_switch[8]] - df["(Q-Qo)/mA.h"].iloc[index_switch[9]]) / A_cell

# capacity fifth charge
capacity_5ch = (df["(Q-Qo)/mA.h"].iloc[index_switch[12]] - df["(Q-Qo)/mA.h"].iloc[index_switch[10]]) / A_cell

# capacity fifth discharge
capacity_5dch = (df["(Q-Qo)/mA.h"].iloc[index_switch[13]] - df["(Q-Qo)/mA.h"].iloc[index_switch[14]]) / A_cell

```

```python
print("Charge capacity cycle 3:", capacity_3ch)
print("Disharge capacity cycle 3:", capacity_3dch)
print("Charge capacity cycle 4:", capacity_4ch)
print("Discharge capacity cycle 4:", capacity_4dch)
print("Charge capacity cycle 5:", capacity_5ch)
print("Discharge capacity cycle 5:", capacity_5dch)

# Coulombic efficiency: discharge capacity / charge capacity
print("Coulombic efficiency cycle 3:", capacity_3dch / capacity_3ch)
print("Coulombic efficiency cycle 4:", capacity_4dch / capacity_4ch)
print("Coulombic efficiency cycle 5:", capacity_5dch / capacity_5ch)
```
