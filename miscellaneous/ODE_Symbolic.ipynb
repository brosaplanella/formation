```python
import sympy as sp

x, t, R, a, L, Js, D = sp.symbols('x t R a L Js D')

# Define the function y(x, t)
y = sp.Function('y')(x, t)

# Define the partial differential equation
eq = sp.Eq(sp.diff(y, x, 2), R * sp.exp(-a * x))

# Define the boundary conditions
bc1 = sp.Eq(sp.diff(y, x).subs(x, 0), 0)
bc2 = sp.Eq(sp.diff(y, x).subs(x, L), L * Js / D)

# Define the initial condition
ic = sp.Eq(y.subs(t, 0), 0)

# Assume a solution in the form of y(x, t) = X(x) * T(t)
X = sp.Function('X')(x)
T = sp.Function('T')(t)
yt = X * T

# Substitute the assumed solution into the PDE
eq_xt = eq.subs(y, yt)
eq_xt = eq_xt.subs(sp.diff(yt, x, 2), sp.diff(X, x, 2) * T)

# Separate the variables and introduce a separation constant λ
eq_sep = sp.simplify(eq_xt.doit())
λ = sp.symbols('λ')
eq_x = sp.Eq(sp.diff(X, x, 2) - λ*X, -R*sp.exp(-a*x)/λ)
eq_t = sp.Eq(sp.diff(T, t), λ*T)

# Solve the spatial equation
sol_x = sp.dsolve(eq_x, X)
print("Solution for X(x):")
sp.pprint(sol_x)

# Apply the boundary conditions to find the values of C₁, C₂, and λ
bc1_x = bc1.subs(y, yt).subs(sp.diff(yt, x), sp.diff(X, x) * T)
bc2_x = bc2.subs(y, yt).subs(sp.diff(yt, x), sp.diff(X, x) * T)

# Solve the temporal equation
sol_t = sp.dsolve(eq_t, T)
print("\nSolution for T(t):")
sp.pprint(sol_t)

# Apply the initial condition to find the value of C₃
ic_xt = ic.subs(y, yt)

# Combine the solutions for X(x) and T(t) to obtain the general solution y(x, t)
# Note: The complete solution will require determining the eigenvalues and eigenfunctions
#       based on the boundary conditions and initial condition.
```
